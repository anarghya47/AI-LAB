```python
def is_variable(expr):

    return isinstance(expr, str) and expr.startswith('?')

def occurs(var, expr, subst):

    if expr == var:
        return True
    if isinstance(expr, str) and expr in subst:
        return occurs(var, subst[expr], subst)
    if isinstance(expr, list):
        return any(occurs(var, arg, subst) for arg in expr[1:])
    return False

def unify(x, y, subst=None):
    if subst is None:
        subst = {}


    while isinstance(x, str) and x in subst:
        x = subst[x]
    while isinstance(y, str) and y in subst:
        y = subst[y]


    if x == y:
        return subst


    if is_variable(x):
        if occurs(x, y, subst):
            return None
        subst[x] = y
        return subst


    if is_variable(y):
        if occurs(y, x, subst):
            return None
        subst[y] = x
        return subst


    if isinstance(x, list) and isinstance(y, list):
        if len(x) != len(y) or x[0] != y[0]:
            return None


        current_subst = dict(subst)
        for a, b in zip(x[1:], y[1:]):
            current_subst = unify(a, b, current_subst)
            if current_subst is None:
                return None
        return current_subst


    return None

def get_expression_input(expr_name):
    print(f"\n--- Input for {expr_name} ---")
    predicate = input("Enter predicate name (e.g., eats, likes): ").strip()

    while True:
        try:
            num_args = int(input("Enter number of arguments: "))
            if num_args >= 0:
                break
            print("Please enter a non-negative integer.")
        except ValueError:
            print("Invalid input. Please enter a valid integer.")

    args = []
    if num_args > 0:
        print("Rule: Prefix variables with a '?' symbol (e.g., ?x, ?person). Plain text means a constant.")
    for i in range(num_args):
        arg = input(f"  Enter argument {i+1}: ").strip()
        args.append(arg)

    return [predicate] + args

def main():
    print("=== Logic Unification Program ===")

    expr1 = get_expression_input("Expression 1")
    expr2 = get_expression_input("Expression 2")

    print(f"\nExpression 1 structured as: {expr1}")
    print(f"Expression 2 structured as: {expr2}")

    result = unify(expr1, expr2)

    print("\n=== Unification Result ===")
    if result is None:
        print("Unification Failed! (A conflict occurred between constants or variable bounds)")
    elif result == {}:
        print("Unification Successful! No variable substitutions were needed.")
    else:
        print("Unification Successful! Substitutions found:")
        for var, val in result.items():
            print(f"  {var} -> {val}")

if __name__ == "__main__":
    main()
```
