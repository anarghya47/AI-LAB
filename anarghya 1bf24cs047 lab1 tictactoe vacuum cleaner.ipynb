```python
b = [" "] * 9

def show():
    print(f"{b[0]}|{b[1]}|{b[2]}")
    print("-+-+-")
    print(f"{b[3]}|{b[4]}|{b[5]}")
    print("-+-+-")
    print(f"{b[6]}|{b[7]}|{b[8]}")

win = [(0,1,2),(3,4,5),(6,7,8),
       (0,3,6),(1,4,7),(2,5,8),
       (0,4,8),(2,4,6)]

p = "X"

for _ in range(9):
    show()
    n = int(input(f"Player {p}, enter position (1-9): ")) - 1

    if b[n] != " ":
        print("Already occupied!")
        continue

    b[n] = p

    if any(b[a] == b[c] == b[d] == p for a,c,d in win):
        show()
        print(p, "wins!")
        break

    p = "O" if p == "X" else "X"
else:
    show()
    print("Draw!")
```

```python

def goal_agent(room, location):
    # Clean the current room if it is dirty
    if room[location] == "Dirty":
        return "Suck"

    # Find the next dirty room
    rooms = list(room.keys())
    current_index = rooms.index(location)

    for i, room_name in enumerate(rooms):
        if room[room_name] == "Dirty":
            if i > current_index:
                return "Move Right"
            elif i < current_index:
                return "Move Left"

    return "Goal Achieved"


# -------------------------------
# Goal-Based Agent
# -------------------------------

num_rooms = int(input("Enter number of rooms: "))

room = {}

# Ask location and status for every room
for i in range(num_rooms):
    location = input(f"\nEnter location of room {i + 1}: ").upper()
    status = input(
        f"Enter status of room {location} (Clean/Dirty): "
    ).capitalize()

    room[location] = status

# Ask the agent's starting location
loc = input("\nEnter agent's starting location: ").upper()

# Goal-based agent
while True:
    action = goal_agent(room, loc)
    print("Action:", action)

    if action == "Goal Achieved":
        break

    elif action == "Suck":
        room[loc] = "Clean"

    elif action == "Move Left":
        rooms = list(room.keys())
        current_index = rooms.index(loc)

        if current_index > 0:
            loc = rooms[current_index - 1]

    elif action == "Move Right":
        rooms = list(room.keys())
        current_index = rooms.index(loc)

        if current_index < num_rooms - 1:
            loc = rooms[current_index + 1]

print("\nAll rooms are clean!")

```
